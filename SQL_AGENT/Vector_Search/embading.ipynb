```python
import requests
import numpy as np
import pandas as pd

OLLAMA_URL = "http://192.168.21.35:11434/api/embeddings"
EMBED_MODEL = "nomic-embed-text"


def get_embedding(text):
    response = requests.post(
        OLLAMA_URL,
        json={
            "model": EMBED_MODEL,
            "prompt": text
        },
        timeout=120
    )

    response.raise_for_status()

    return response.json()["embedding"]


def cosine_similarity(a, b):
    a = np.array(a, dtype=np.float32)
    b = np.array(b, dtype=np.float32)

    return np.dot(a, b) / (
        np.linalg.norm(a) * np.linalg.norm(b)
    )


def search_tables(question, top_k=10):

    # --------------------------------------------------
    # 1. Embed the user's question
    # --------------------------------------------------

    query_embedding = get_embedding(question)

    results = []

    # --------------------------------------------------
    # 2. Compare against every business meaning
    # --------------------------------------------------

    for _, row in df.iterrows():

        table_embedding = row["embeddings"]

        similarity = cosine_similarity(
            query_embedding,
            table_embedding
        )

        results.append({
                **row.to_dict(),
                "similarity": similarity
            })

    # --------------------------------------------------
    # 3. Rank highest similarity first
    # --------------------------------------------------

    results_df = pd.DataFrame(results)

    results_df = results_df.sort_values(
        "similarity",
        ascending=False
    ).reset_index(drop=True)

    return results_df.head(top_k)
```

```python
from collections import defaultdict, deque


def build_fk_graph(kb):

    graph = defaultdict(list)

    for table in kb:

        table_name = (
            table.get("physical_qualified_name")
            or table.get("qualified_name")
        )

        if not table_name:
            continue

        for fk in table.get("foreign_keys", []):

            fk_column = fk.get("column")
            ref_table = fk.get("references_table")
            ref_column = fk.get("references_column")

            if not fk_column or not ref_table or not ref_column:
                continue

            graph[table_name.lower()].append({
                "from_table": table_name,
                "from_column": fk_column,
                "to_table": ref_table,
                "to_column": ref_column,
                "direction": "forward"
            })

            # Add reverse edge
            graph[ref_table.lower()].append({
                "from_table": ref_table,
                "from_column": ref_column,
                "to_table": table_name,
                "to_column": fk_column,
                "direction": "reverse"
            })

    return graph
```

```python
def trace_relationships(
    start_tables,
    fk_graph,
    max_depth=3
):

    visited = set()
    queue = deque()

    discovered = []

    for table in start_tables:

        key = table.lower()

        if key not in visited:

            visited.add(key)

            queue.append(
                (table, 0)
            )

    while queue:

        current_table, depth = queue.popleft()

        discovered.append({
            "table": current_table,
            "depth": depth
        })

        if depth >= max_depth:
            continue

        current_key = current_table.lower()

        for edge in fk_graph.get(
            current_key,
            []
        ):

            next_table = edge["to_table"]

            next_key = next_table.lower()

            if next_key not in visited:

                visited.add(next_key)

                queue.append(
                    (
                        next_table,
                        depth + 1
                    )
                )

    return discovered
```

```python
def get_relevant_tables_from_question(
    question,
    related_tables,
    semantic_kb
):

    question_lower = question.lower()

    relevant = []

    for table in semantic_kb:

        table_name = (
            table.get("physical_qualified_name")
            or table.get("qualified_name")
            or ""
        )

        if table_name.lower() not in [
            t.lower() for t in related_tables
        ]:
            continue

        text = " ".join([
            table_name,
            table.get("table_relationship_role", ""),
            " ".join(table.get("useful_join_paths", []))
        ]).lower()

        # Very simple initial matching
        question_words = question_lower.split()

        score = sum(
            1
            for word in question_words
            if len(word) > 2 and word in text
        )

        if score > 0:
            relevant.append({
                "table": table_name,
                "score": score
            })

    relevant.sort(
        key=lambda x: x["score"],
        reverse=True
    )

    return relevant
```

```python
def build_qwen_schema_context(
    candidate_tables,
    semantic_kb
):

    lookup = {}

    for table in semantic_kb:

        name = (
            table.get("physical_qualified_name")
            or table.get("qualified_name")
        )

        if name:
            lookup[name.lower()] = table

    context = []

    for item in candidate_tables:

        table_name = item["table"]
        score = item["score"]

        table = lookup.get(
            table_name.lower()
        )

        if not table:
            continue

        context.append({
            "table": table_name,
            "embedding_score": score,
            "schema": table
        })

    return json.dumps(
        context,
        indent=2,
        ensure_ascii=False
    )
```

## Prototype: hybrid semantic + lexical anchor retrieval

Reuses the finalized graph/bridge/lookup/context machinery from
`lms_chatbot.retrieval` (already built and in production) and only
prototypes the one new piece we're testing here: **anchor selection**
that combines embedding similarity (semantic, always on - handles
paraphrased questions with no literal table/column names) with a
lexical term-overlap score (a boost/rescue signal for when the user's
words happen to match a table name, business term, or column - never
a requirement, since we can't expect users to know exact entity names).

If this looks right across the test questions below, this scoring
function moves into `lms_chatbot/retrieval/embeddings.py` and
`pipeline.py` gets updated to call it instead of pure cosine ranking.

```python
import sys

REPO_ROOT = base.parent  # d:\P_PC_CHATBOT
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

from lms_chatbot.retrieval import classify, context as context_mod, graph as graph_mod
from lms_chatbot.retrieval.embeddings import cosine_similarity, get_embedder
from lms_chatbot.retrieval.build_index import _embedding_text

print("Imported production retrieval modules from:", REPO_ROOT)
```

```python
# Reload the FULL kb fresh - the `semantic_kb` variable above was stripped
# down to only a few fields earlier in this notebook (cell 63b24350) and
# no longer has business_terms/columns/etc. we need for lexical matching.
with open(SEMANTIC_KB_PATH, "r", encoding="utf-8") as f:
    full_kb = json.load(f)

anchors, lookups = classify.split_tables(full_kb)

kb_lookup = {}
for table in full_kb:
    name = classify.table_name(table)
    if name:
        kb_lookup[name.lower()] = table

name_index = classify.NameIndex(full_kb)
fk_graph = graph_mod.build_fk_graph(full_kb, name_index)

print(f"Loaded {len(full_kb)} tables total: {len(anchors)} anchors, {len(lookups)} lookups")
```

```python
# Embed every anchor's business description ONCE (this is what
# lms_chatbot.retrieval.build_index does as an offline step). Query time
# below only ever embeds the *question*, never re-embeds table text -
# fixing the per-query re-embed the earlier cells in this notebook did.
embedder = get_embedder()

anchor_names = []
anchor_texts = []

for table in anchors:
    text = _embedding_text(table)
    if text:
        anchor_names.append(classify.table_name(table))
        anchor_texts.append(text)

anchor_vectors = dict(zip(anchor_names, embedder.embed_documents(anchor_texts)))

print(f"Embedded {len(anchor_vectors)} anchor tables")
```

```python
import re

_STOPWORDS = {
    "the", "a", "an", "is", "are", "was", "were", "and", "or", "of", "in",
    "on", "to", "for", "by", "with", "all", "list", "show", "get", "find",
    "what", "which", "who", "that", "this", "these", "those", "me",
    "please", "give", "display", "tell", "want", "need", "can", "you",
    "does", "do", "did", "has", "have", "had", "not", "any",
}


def tokenize(text: str) -> set[str]:
    words = re.findall(r"[a-z0-9]+", text.lower())
    return {w for w in words if len(w) > 2 and w not in _STOPWORDS}


def build_lexical_bag(table: dict) -> set[str]:
    """Bag of literal terms a user might type that should point at this table -
    table/column names, business terms, aliases, column synonyms - as opposed
    to embedding text, which is full descriptive prose for semantic matching."""

    parts = []

    for key in ("business_name", "table_name", "qualified_name", "physical_qualified_name"):
        value = table.get(key)
        if value:
            parts.append(value)

    parts.extend(table.get("business_terms") or [])
    parts.extend(table.get("aliases") or [])
    parts.extend(table.get("related_business_concepts") or [])
    parts.extend(table.get("likely_user_phrases") or [])

    for column in table.get("columns") or []:
        if column.get("column_name"):
            parts.append(column["column_name"])
        parts.extend(column.get("synonyms") or [])
        parts.extend(column.get("likely_user_terms") or [])

    bag: set[str] = set()
    for part in parts:
        bag |= tokenize(part)

    return bag


lexical_index = {
    classify.table_name(table): build_lexical_bag(table)
    for table in anchors
    if classify.table_name(table)
}

print(f"Built lexical bags for {len(lexical_index)} anchor tables")
```

```python
# Semantic is the primary signal and always runs (handles paraphrases with
# zero literal table/column names). Lexical is a BOOST, never a gate: a
# table with no lexical overlap at all can still win purely on semantic
# similarity; lexical only pushes a table up (or in) when the user's words
# happen to match its name/terms/columns.
LEXICAL_BOOST = 0.5


def lexical_score(question_tokens: set[str], bag: set[str]) -> float:
    if not question_tokens or not bag:
        return 0.0
    return len(question_tokens & bag) / len(question_tokens)


def hybrid_anchor_scores(question: str, top_n: int):
    question_tokens = tokenize(question)
    query_vector = embedder.embed_query(question)

    scored = []
    for name, vector in anchor_vectors.items():
        semantic = cosine_similarity(query_vector, vector)
        lexical = lexical_score(question_tokens, lexical_index.get(name, set()))
        final = semantic + LEXICAL_BOOST * lexical
        scored.append({
            "table": name,
            "final": final,
            "semantic": semantic,
            "lexical": lexical,
        })

    scored.sort(key=lambda row: row["final"], reverse=True)
    return scored[:top_n]


def get_relevant_schema_prototype(question, top_n=6, max_tables=14, bridge_max_hops=4):
    scored_anchors = hybrid_anchor_scores(question, top_n)
    anchor_names_selected = [row["table"] for row in scored_anchors]

    bridges = sorted(
        graph_mod.bridge_tables(fk_graph, anchor_names_selected, bridge_max_hops)
        - {a.lower() for a in anchor_names_selected}
    )

    core = anchor_names_selected + bridges
    lookup_names = sorted(
        graph_mod.attach_lookups(kb_lookup, name_index, core, classify.is_lookup_table)
        - {c.lower() for c in core}
    )

    selected = (
        [(name, "anchor") for name in anchor_names_selected]
        + [(name, "bridge") for name in bridges]
        + [(name, "lookup") for name in lookup_names]
    )[:max_tables]

    tables_with_roles = []
    for name, role in selected:
        entry = kb_lookup.get(name.lower())
        if entry:
            tables_with_roles.append((entry, role))

    return {
        "question": question,
        "scored_anchors": scored_anchors,
        "anchors": anchor_names_selected,
        "bridges": bridges,
        "lookups": lookup_names,
        "total_tables": len(tables_with_roles),
        "context": context_mod.render_context(tables_with_roles, name_index),
    }
```

```python
test_questions = [
    # 1. Pure paraphrase, no literal table/column names - semantic must carry this alone.
    "who has completed their training",
    # 2. Literal table name mention - lexical should boost/rescue this.
    "show me tm_attachment files",
    # 3. Multi-entity question - two anchors should surface, bridged correctly.
    "list all trainings and their categories",
    # 4. Different domain (masters.*, not lms.*) - checks anchors stay scoped,
    #    no leakage into unrelated hub tables (gender, qr_code, language, ...).
    "which security group does a user belong to",
]

for q in test_questions:
    result = get_relevant_schema_prototype(q)

    print("=" * 100)
    print("Q:", result["question"])
    print(f"-> {result['total_tables']} tables total "
          f"({len(result['anchors'])} anchor, {len(result['bridges'])} bridge, {len(result['lookups'])} lookup)")
    print()
    print("Anchor scores (table | final | semantic | lexical):")
    for row in result["scored_anchors"]:
        print(f"  {row['table']:<30} {row['final']:.3f}   {row['semantic']:.3f}   {row['lexical']:.3f}")
    print()
    print("Bridges:", result["bridges"] or "(none)")
    print("Lookups:", result["lookups"] or "(none)")
    print()
    print(result["context"])
    print()
```
